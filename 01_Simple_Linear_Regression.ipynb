# Practical 1: Simple Linear Regression

## Aim
To build a machine learning model that predicts a student's marks based on the number of hours studied.

## Theory
Linear regression is a supervised machine learning algorithm used to predict a numerical value.

In this practical:
- Input feature: Hours studied
- Target variable: Marks obtained
- Algorithm: Linear Regression
- Evaluation: Mean Absolute Error (MAE)

## Dataset
We will create a small sample dataset for learning purposes.


## Step 1: Import the required Python libraries

```python

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

print("Libraries imported successfully!")
```

## Step 2: Create The dataset

```python

data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [35, 40, 50, 55, 65, 70, 80, 85]
}

df = pd.DataFrame(data)

print(df)
```

# Step 3: Visualize the dataset

```python

plt.scatter(df["Hours_Studied"], df["Marks"])

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.show()
```

# Step 4: Split the data into training and testing sets

```python

X = df[["Hours_Studied"]]
y = df["Marks"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

print("Training examples:", len(X_train))
print("Testing examples:", len(X_test))
```

# Step 5: Train the machine learning model

```python

model = LinearRegression()

model.fit(X_train, y_train)

print("Model training completed!")
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

# Step 6: Predict Marks

```python
predictions = model.predict(X_test)

results = pd.DataFrame({
    "Actual_Marks": y_test.to_numpy(),
    "Predicted_Marks": predictions
})

print(results.round(2))
```

# Step 7: Evaluate the model

```python

mae = mean_absolute_error(y_test, predictions)

print("Mean Absolute Error:", round(mae, 2))
```

# Step 8: Predict for a new student

```python

new_student = pd.DataFrame({
    "Hours_Studied": [6.5]
})

predicted_marks = model.predict(new_student)

print(
    "Predicted marks for 6.5 study hours:",
    round(predicted_marks[0], 2)
)
```

# Step 9: Plot the regression line

```python

plt.scatter(
    X_train["Hours_Studied"],
    y_train,
    label="Training data"
)

plt.plot(
    X_train["Hours_Studied"],
    model.predict(X_train),
    label="Regression line"
)

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression Model")
plt.legend()
plt.show()
```
